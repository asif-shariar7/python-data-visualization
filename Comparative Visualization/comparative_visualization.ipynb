<h5>Comparative Visualization</h5>

<p>
    <b>Comparative visualization</b> is a technique of plotting 
    <b>two or more variables or groups together</b> in a single chart 
    so their <b>differences or patterns</b> can be directly compared, 
    instead of studying each variable in isolation.
</p>


<p>Why do we use comparative visualization?</p>

<p>
    When there are many numbers in a table, it can be difficult to identify
    differences quickly.
</p>

<p>For example :</p>

<table>
    <tr>
        <th>City</th>
        <th>Count</th>
    </tr>
    <tr>
        <td>Dhaka</td>
        <td>90</td>
    </tr>
    <tr>
        <td>Jashore</td>
        <td>97</td>
    </tr>
    <tr>
        <td>Khulna</td>
        <td>83</td>
    </tr>
    <tr>
        <td>Rajshahi</td>
        <td>83</td>
    </tr>
    <tr>
        <td>Faridpur</td>
        <td>72</td>
    </tr>
    <tr>
        <td>Mymensingh</td>
        <td>80</td>
    </tr>
</table>

<p>
    we can read the numbers, but a visualization lets our eyes compare
    them immediately.
</p>

<p>So, comparative visualization helps us:</p>

<ol>
    <li><b>Identify differences</b></li>
    <li><b>Rank categories</b></li>
    <li><b>Find the highest and lowest values</b></li>
    <li><b>Compare groups</b></li>
    <li><b>Find patterns</b></li>
    <li><b>Communicate results clearly</b></li>
    <li><b>Make data-driven decisions</b></li>
</ol>


<p><b>Common types of comparative visualization</b></p>

<p>
    Different charts are useful for different types of comparisons.
</p>


<p><b>1. Bar Chart</b></p>

<p><b>2. Grouped Bar Chart</b></p>

<p>

</p>

<p>

    </tr>

        <td>100</td>

    <tr>

        <td>150</td>

    <tr>

    </tr>

<p>

<ul>

<p><b>3. Comparative Line Chart</b></p>

<p>
    A <b>line chart</b> is useful when we want to compare
    <b>changes over time</b>.
</p>

<p>
    For example, we can compare the monthly sales of two companies.
</p>

<ul>
    <li><b>Company A vs Company B</b></li>
    <li>Their growth over time</li>
    <li>Which company consistently has higher sales</li>
    <li>Whether the gap is increasing or decreasing</li>
</ul>

<p>
    <b>Important:</b> When the main purpose is a <b>trend over time</b>,
    a line chart is generally better than a bar chart.
</p>

<p><b>Important Rules for Comparative Visualization</b></p>

<p>Rule 1: Choose the chart based on the data</p>

<p>
    Do not choose a chart just because it looks attractive.
</p>

<ul>
    <li>Categories - <b>Bar chart</b></li>
    <li>Time - <b>Line chart</b></li>
    <li>Two numerical variables - <b>Scatter plot</b></li>
    <li>Parts of one whole - <b>Pie/stacked bar</b></li>
</ul>


<p>Rule 2: Make comparisons easy</p>

<p>
    The viewer should be able to understand the difference quickly.
</p>

<p>
    For example:
</p>

<pre>
Product A - 100
Product B - 150
Product C - 80
</pre>

<p>
    A bar chart immediately shows:
    <b>B &gt; A &gt; C</b>
</p>


<p>Rule 3: Use the same scale (the range and intervals of numbers shown on a chart axis)</p>

<p>
    When comparing multiple charts or groups, the scale should be consistent
    whenever possible.
</p>

<p>
    For example, do not make one chart use <b>0 - 100</b> and another use
    <b>80 - 100</b> without clearly indicating the difference.
</p>

<p>
    This can make a small difference look much larger than it really is.
</p>


<p>Rule 4: Label the chart clearly</p>

<p>
    A good comparative visualization should normally have:
</p>

<ul>
    <li><b>Title</b></li>
    <li><b>Category labels</b></li>
    <li><b>Value/axis labels</b></li>
    <li><b>Legend</b>, when multiple groups are shown</li>
</ul>

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
df.describe()
```

```python
plt.figure(figsize = (10,6))
sns.countplot( data = df, x = 'class', hue = 'sex')
plt.show()
```

```python
# Compare passenger distribution by gender and class
plt.figure(figsize = (10,7))
ax = sns.countplot( data = df, x = 'class', hue = 'sex')
plt.title("Passenger class Distribution by Gender")
plt.ylabel("count")


for i in ax.containers:
    ax.bar_label(i, padding = 3, fontsize = 9)


plt.show()
```

<p>Checking - </p>

```python
df['sex'].value_counts()
```

```python
# compare survival by gender and class
plt.figure(figsize=(10,7))
ax = sns.countplot(data=df, x='class', hue='survived')
plt.title("Survival by Passenger Class")
plt.ylabel("Count")

for i in ax.containers:
    ax.bar_label(i, padding=3, fontsize=9)

plt.show()
```

<p>Checking - </p>

```python
df['survived'].value_counts()
```

```python
# Age distribution by survival

plt.figure(figsize=(10,6))
sns.boxplot(x='survived', y='age', data=df)

plt.title("Age Distribution: Survived vs Not Survived")
plt.xlabel("Survival (0 = No, 1 = Yes)")
plt.ylabel("Age")

plt.show()
```
