<h5>Text Visualization</h5>
<p>Text visualization means showing text data in a way that makes it easier to understand, analyze, and find patterns. In Python, text can be visualized using different methods. One common method is : </p>

<p><b>Word Cloud</b></p>

<p>A word cloud shows frequently used words in larger sizes.</p>
<ul>
  <li>More frequent word - larger</li>
  <li>Less frequent word - smaller</li>
</ul>

<p>Example uses : </p>

<ul>
  <li>Customer reviews</li>
  <li>News articles</li>
  <li>Social media comments</li>
</ul>

<p>It helps us quickly understand the most common or popular words in a collection of text.</p>

<p>Syntax - </p>

```bash

```

````python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from wordcloud import WordCloud
import matplotlib.pyplot as plt
````

<p>Example of Word Cloud - </p>

````python
text = """
Python is useful for data analysis.
Python is also useful for machine learning.
Data analysis helps us understand data.
Python is easy to learn.
"""

wordcloud = WordCloud().generate(text)

plt.imshow(wordcloud)  # displays the Word Cloud as an image.
plt.axis('off')        # hides the x-axis and y-axis.

plt.show()
````

<p>Here,
<ul>
    <li><b>WordCloud()</b> - creates a Word Cloud object.</li>
    <li><b>.generate(text)</b> - takes the text and generates the Word Cloud based on word frequency.</li>
    <li><b>wordcloud = </b> - stores the generated Word Cloud in the <b>wordcloud</b> variable.</li>
</ul>

<h5>Vectorizer</h5>

<p>A <b>vectorizer</b> converts text documents into <b>numbers</b> so that a computer can analyze the text. This process is called <b>vectorization</b>.</p>

<h5>Text Visualization Steps</h5>

<ol>
    <li>Get Data</li>
    <li>Check data</li>
    <li>Select Text</li>
    <li>Create Vectorizer</li>
    <li>Convert Text -> Numbers</li>
    <li>Calculate Word Frequency</li>
    <li>Create WordCloud -> Display</li>
</ol>

````python
newsgroups = fetch_20newsgroups(subset='train', categories=['sci.space'])
text = newsgroups.data[:100]
````

<hr>

````python
vectorizer = CountVectorizer(stop_words = "english", max_features = 1000)
X = vectorizer.fit_transform(text)
wordfrq = dict(zip(vectorizer.get_feature_names_out(), X.toarray().sum(axis = 0)))
````

````python
wordcloud = WordCloud(width=800, height=400, background_color="white").generate_from_frequencies(wordfrq)
plt.figure(figsize = (12,6))
# plt.imshow(wordcloud)
plt.imshow(wordcloud, interpolation = 'lanczos')
plt.axis("off")
plt.title("Top words in news articles")
plt.show()
````

<p>Here, <b>CountVectorizer</b> : For text vectorization using <b>CountVectorizer</b>, we use the <b>Scikit-learn (sklearn)</b> library.</p>

```bash
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(stop_words='english',max_features=1000)
```

<p>What does it do?</p>

<p><code>CountVectorizer</code>:</p>

<ul>
    <li>Converts text into a <b>matrix of word counts</b>.</li>
    <li>Removes common English words such as <code>the</code>, <code>is</code>, and <code>and</code>.</li>
    <li>Keeps the <b>top 1000 most frequent words</b>.</li>
</ul>

<p>Example</p>

<p>Suppose we have:</p>

```bash

Data is useful
```

<p>The vectorizer counts how many times each important word appears.</p>

```bash

```

<p>This converts the text documents into a <b>numerical matrix</b>.</p>

<h5>How to Check if Data is Suitable for <b>CountVectorizer</b></h5>

<p>
    Before using <b>CountVectorizer</b>, first check whether the dataset
    contains <b>raw text documents</b>.
</p>

<p>Step 1: Load the Dataset</p>

<p>Step 2: Check the Dataset Type</p>

```bash
print(type(data))
```

<p>
    This tells us what type of dataset object we have.
</p>

<p>Step 3: Check What is Inside the Dataset</p>

```bash
print(data.keys())
```

<p>Step 4: Check the Data Type</p>

```bash
print(type(data.data))
```

<p>
    This tells us how the documents are stored. A <code>list</code> alone does
    <b>not</b> mean that the data is text.
</p>

<p>Step 5: Check One Data Item</p>

<p>
    This is the <b>most important step</b>.
</p>

```bash
print(type(data.data[0]))
print(data.data[0])
```

<p>Step 6: Decide if it is Raw Text</p>

<table>
<tr>
    <th>What We See</th>
    <th>Data Type</th>
    <th>CountVectorizer?</th>
</tr>

<tr>
    <td>Sentence, paragraph, article, review, or document</td>
    <td>Raw Text</td>
    <td><b>Yes </b></td>
</tr>

<tr>
    <td>Numbers or numerical array</td>
    <td>Numerical Data</td>
    <td><b>No </b></td>
</tr>

<tr>
    <td>Pixel values / image array</td>
    <td>Image Data</td>
    <td><b>No </b></td>
</tr>

<tr>
    <td>Sparse matrix / numerical vectors</td>
    <td>Already Vectorized</td>
    <td><b>No </b></td>
</tr>
</table>

<p>Example: Raw Text </p>

```bash
print(data.data[0])
```

<p>Output may look like:</p>

```bash
NASA has launched a new spacecraft.
The spacecraft will travel through space...
```

<p>
    This is an actual text document, so <b>CountVectorizer</b> can be used.
</p>


<p>Example: Numerical Data </p>

```bash
[8.3252, 41.0, 6.9841, 1.0238, ...]
```

<p>
    This contains numbers, not text documents. Therefore,
    <b>CountVectorizer</b> should not be used.
</p>

<p>Example: Image Data </p>

```bash
[[0, 0, 255],
 [128, 255, 0],
 [255, 255, 255]]
 ```

<p>
    These are pixel values, not text. Therefore,
    <b>CountVectorizer</b> should not be used.
</p>

<p>Example: Already Vectorized Data </p>

<p>
    If the dataset already contains numerical vectors or a sparse matrix,
    the text has already been converted into numbers. Therefore,
    <b>CountVectorizer</b> should not be applied again.
</p>



<p>
    So, If the output is an actual <b>sentence, paragraph, article, review, or
    document</b> - <b>Raw Text - CountVectorizer </b>
</p>

<p>
    If the output is <b>numbers, image pixels, or already-vectorized values</b>
    - <b>Do not use CountVectorizer </b>
</p>

<p>
    <b>So, Load Dataset -> Find Data -> Check One Sample -> Is It Raw Text?</b>
</p>

<p>
    <b>Yes - CountVectorizer</b><br>
    <b>No - Don't use CountVectorizer</b>
</p>
