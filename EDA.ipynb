<h3>EDA (Exploratory Data Analysis)</h3>

<p><b>EDA</b> stands for <b>Exploratory Data Analysis</b>. It is the process of <b>examining, cleaning, summarizing, and visualizing data</b> to understand its important features and find patterns or problems.</p>

<h4>Main Purposes of EDA</h4>

<ul>
    <li>Understand the data</li>
    <li>Find missing values</li>
    <li>Detect outliers</li>
    <li>Find patterns and relationships</li>
    <li>Understand data distribution</li>
    <li>Prepare data for further analysis or machine learning</li>
</ul>

<h4>Common Python Libraries for EDA</h4>

<ul>
    <li><b>Pandas</b> → Data cleaning and analysis</li>
    <li><b>NumPy</b> → Numerical operations</li>
    <li><b>Matplotlib</b> → Data visualization</li>
    <li><b>Seaborn</b> → Statistical visualization</li>
</ul>

<h4>Common EDA Steps</h4>

<ol>
    <li>Load the dataset</li>
    <li>Check the data structure</li>
    <li>Check missing values</li>
    <li>Remove or handle duplicates</li>
    <li>Analyze statistical summary</li>
    <li>Detect outliers</li>
    <li>Visualize the data</li>
    <li>Find relationships between variables</li>
</ol>

<h4>Example - </h4>

<p>
    Suppose we have student data containing 
    <b>Name, Age, Marks,</b> and <b>Attendance</b>.
</p>

<p>Using EDA, we can find:</p>

<ul>
    <li>Average marks</li>
    <li>Highest and lowest marks</li>
    <li>Students with missing data</li>
    <li>Relationship between attendance and marks</li>
    <li>Distribution of marks</li>
</ul>

<blockquote>
    <b>EDA is the process of understanding and exploring a dataset before performing
    deeper analysis or building a machine learning model.</b>
</blockquote>

````python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
````

<h5>Data Generation - </p>

````python
towns = ['Jashore', 'Mymensingh', 'Faridpur', 'Khulna', 'Rajshahi', 'Dhaka']
complaint_types = ['Power Cut', 'Traffic', 'Trafick', 'Noise', 'Garbage', 'Water Leak', 'Power Cut', 'Ice Cream Cart']
ice_cream_flavors = ['Vanilla', 'Chocolate', 'Strawberry', 'Mint', 'Mango', 'Lemon', 'Coffee']

n_rows = 500

data = {
    'towns': np.random.choice(towns, size=n_rows),
    'complaint_type': np.random.choice(complaint_types, size=n_rows),
    'temperature': np.random.normal(loc=32, scale=5, size=n_rows),
    'electricity_usage': np.random.normal(loc=250, scale=50, size=n_rows),
    'num_ice_cream_carts': np.random.choice([0, 1, 2, 5, 10, 10, 10], size=n_rows),  # baisness create (occurs 10 more)
    'survey_rating': np.random.choice([1, 2, 3, 4, 5, np.nan], size=n_rows),
}

df = pd.DataFrame(data)
````

<p>Here, <b>np.random.choice()</b> is a NumPy function used to randomly select one or more values from an array or range.</p>
<p>Syntax - </p>

```bash
np.random.choice(a, size=None, replace=True, p=None)
```
<p>Here, <br>
a - Array or integer to choose from. If an integer n is given, it chooses from 0 to n-1.<br>
size - Number of random values to return.<br>
replace	True = values can repeat (default). False = no repeated values.<br>
p- Probability of selecting each value.</p>

<p>Example - </p>

````python
# colors = ["Red", "Blue", "Green"]
# print(np.random.choice(colors, size = 10,  p = [0.6, 0.3, 0.1]))
````

<p>Here, <b>np.random.normal()</b> Used to generate random numbers from a normal (Gaussian) distribution.</p>

```bash
np.random.normal(loc=0, scale=1, size=None)
```
<p>Here, <br>
loc - Mean (default = 0). Suppose,  loc = 50 - Average value is 50.<br>
scale - Standard deviation (default = 1). Suppose, scale = 7 - Values are usually close to 50, but not limited to 50 ± 7<br>
size - Number or shape of values</p>

<h5>Checking Category Imbalance Using <b>value_counts()</b></h5>
<p>The <b>value_counts()</b> function is used to count how many times each unique value appears in a column. It can help us identify <b>category imbalance</b>, which may indicate a possible <b>representation bias</b> in the dataset.</p>

<p>Example:</p>

```bash
print(df['towns'].value_counts())
```

<p>Example Output:<br>
Dhaka - 800<br>
Chittagong - 150<br>
Rajshahi - 50<br>

This output shows that the dataset contains:<br>
800 records from Dhaka<br>
150 records from Chittagong<br>
50 records from Rajshahi<br>

This means the data is imbalanced across towns.<br>
If one town has many more records than others, our analysis or machine learning model may represent that town more strongly than the others.<br>

<p>Note: <b>value_counts()</b> does not directly prove bias. It only helps us identify category imbalance or unequal representation.
Category imbalance - May indicate possible representation bias - Requires further investigation</p>

We can use a Seaborn <b>countplot()</b> to visualize the imbalance.<br>

```bash
sns.countplot(data=df, x='towns')
plt.xticks(rotation=45)
plt.show()
```

value_counts() - Counts each category<br>
countplot() - Visualizes category counts<br>
Both can help us detect possible imbalance or representation issues in categorical data.</p>

````python
print(df['towns'].value_counts())   # it can be used to check for possible bias
````

````python
sns.set(style="whitegrid")
plt.figure(figsize=(8,4))
sns.histplot(df['temperature'], kde=True, bins=10)
plt.title("Temperature distribution")
plt.xlabel("temperature degree")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
````

<p>Here, <b>sns.set()</b> is used to set the overall style and appearance of Seaborn plots.</p>

<p>Common Styles - </p>

<ul>
  <li><code>darkgrid</code> - Dark background with grid</li>
  <li><code>whitegrid</code> - White background with grid</li>
  <li><code>dark</code> - Dark background</li>
  <li><code>white</code> - White background</li>
  <li><code>ticks</code> - White background with axis ticks</li>
</ul>

````python
plt.figure(figsize=(10,6))
sns.boxplot(x='towns', y='electricity_usage', data=df)
plt.title("Electricity usage by towns")
plt.xlabel("Town")
plt.ylabel("electricity usage")
plt.tight_layout()
plt.show()
````

````python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="plasma", fmt = "0.2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
````
