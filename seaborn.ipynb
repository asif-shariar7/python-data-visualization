<p>Many Seaborn functions use similar parameters - </p>

```bash 
sns.scatterplot(
    data=DataFrame,   
    x="Column for the X-axis.",
    y="Column for the Y-axis.",
    hue="categorical_column",   # Colors points by category
    style="categorical_column", # Uses different marker shapes
    size="numeric_column",      # Changes point size based on values
    palette="palette_name",     # Chooses the color theme
    color="color_name",         # Makes all points one color
    alpha=0.5,                  # Makes points semi-transparent
    legend=True                 # Shows the legend
)
```
<p>Here, </p>
<p>hue - Uses color to represent another variable.</p>
<pre><code>hue="gender"</code></pre>
<p>Example:</p>
<ul>
  <li>Male - Blue</li>
  <li>Female - Orange</li>
</ul>

<p>style - Uses different marker shapes.</p>
<pre><code>style="department"</code></pre>
<p>Example:</p>
<ul>
  <li>IT - Circle</li>
  <li>HR - Triangle</li>
</ul>

<p>size - Changes marker size based on numeric values.</p>
<pre><code>size="experience"</code></pre>
<p>Example:</p>
<ul>
  <li>1 year - Small dot</li>
  <li>5 years - Medium dot</li>
  <li>10 years - Large dot</li>
</ul>

<p>palette - Chooses the color palette.</p>
<pre><code>palette="Set2"</code></pre>
<p>Example:</p>
<ul>
  <li><code>Set1</code> - Bright colors</li>
  <li><code>Pastel1</code> - Light pastel colors</li>
</ul>
<p>Common built-in palettes:</p>

<h5>Lineplot</h5>

````python
df = sns.load_dataset('titanic')
df
````

<p>A line plot is used to show how a value changes over time or in order by connecting data points with lines.</p>
<p>X-axis: Can be numeric or categorical (ordered)<br>
Y-axis: Usually numeric</p>

<p>Use line plots to find:</p>
<ul>
    <li>Trends</li>
    <li>Increases or decreases</li>
    <li>Changes over time</li>
    <li>Seasonal patterns</li>
    <li>Compare trends between categories (using <code>hue</code>)</li>
</ul>

````python
# sns.lineplot(data = df, x = 'age', y = 'fare', errorbar = None)
sns.lineplot(data = df, x = 'age', y = 'fare')
plt.show()
````

<p>How to read this graph:</p>

<ul>
    <li><b>X-axis:</b> passenger age. <b>Y-axis:</b> fare paid.</li>
    <li>The <b>dark blue line</b> shows the average fare for each age.</li>
    <li>The <b>light blue band</b> shows how sure we are about that average. A wide band means very few passengers of that age, so it's not very reliable. A narrow band means more passengers, so it's more reliable.</li>
</ul>

<p>What it tells us:</p>

<ul>
    <li>Fare doesn't rise smoothly with age - it jumps around a lot. That's because fare depends more on things like ticket class or family size, not really on age.</li>
    <li>Very young passengers (0–5) and older ones (60–65) show big spikes in fare, but the band is wide there too, meaning very few people that age, so we shouldn't trust those spikes much.</li>
    <li>Ages 20–50 have a narrower band, meaning more passengers, so those numbers are more trustworthy — fares here mostly stay between 20 and 50.</li>
</ul>

<p><b>Note:</b> age alone doesn't clearly predict fare, the pattern is messy, and the shaded band tells you how much to trust each part of the line.</p>

<h5>Heatmap</h5>

<p>A heatmap represents numerical values using different colors. It is commonly used to visualize a correlation matrix.</p>

<p>Example - </p>
<p>First, calculate the correlation:</p>

```bash
corr = df.corr(numeric_only=True)
```

<p>Then create the heatmap:</p>

```bash
sns.heatmap(corr)
plt.show()
```

<p>The <b>annot=True</b> parameter displays the numerical values inside the cells.</p>

```bash
sns.heatmap(corr, annot=True)
plt.show()
```

<p>We can use <b>cmap</b> to control the color scheme.</p>

```bash
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
```

<p>Common <b>cmap</b> Values - </p>

<ul>
    <li><code>"Blues"</code></li>
    <li><code>"Greens"</code></li>
    <li><code>"Reds"</code></li>
    <li><code>"Oranges"</code></li>
    <li><code>"Purples"</code></li>
    <li><code>"YlOrRd"</code></li>
    <li><code>"viridis"</code></li>
    <li><code>"magma"</code></li>
    <li><code>"plasma"</code></li>
    <li><code>"cividis"</code></li>
    <li><code>"coolwarm"</code></li>
    <li><code>"RdBu"</code></li>
    <li><code>"RdYlBu"</code></li>
    <li><code>"PiYG"</code></li>
    <li><code>"Spectral"</code></li>
</ul>

<p>For correlation heatmaps, <code>"coolwarm"</code> is one of the most commonly used choices.</p>

<p>Common Parameters - </p>
<ul>
    <li><b>annot</b> - Displays values inside cells</li>
    <li><b>cmap</b> - Controls the color map</li>
    <li><b>linewidths</b> - Adds lines between cells</li>
    <li><b>fmt</b> - Controls the number format of annotations</li>
</ul>

<p>Example with Formatting</p>

```bash
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm",linewidths=0.5)
plt.show()
```

<p><b>fmt=".2f"</b> displays values with 2 decimal places.</p>

<p>Understanding Correlation Values - </p>
<p>Correlation values usually range from <b>-1 to +1</b>.</p>
<ul>
    <li><b>+1</b> - Perfect positive relationship</li>
    <li><b>0</b> - No linear relationship</li>
    <li><b>-1</b> - Perfect negative relationship</li>
</ul>

<p>For example:</p>

<ul>
    <li><b>+0.90</b> - Strong positive correlation</li>
    <li><b>+0.10</b> - Very weak positive correlation</li>
    <li><b>0.00</b> - No linear correlation</li>
    <li><b>-0.10</b> - Very weak negative correlation</li>
    <li><b>-0.90</b> - Strong negative correlation</li>
</ul>

<p>Use a heatmap to:</p>
<ul>
    <li>Visualize correlation matrices</li>
    <li>Find relationships between numerical variables</li>
    <li>Identify highly correlated variables</li>
    <li>Understand patterns in matrix data</li>
</ul>

<p>Note: Correlation does not mean causation. It means two variables can be related, but one variable does not necessarily cause the other.</p>

<p>Example:</p>
<p>Study time increases - Exam scores increase</p>
<p>This does not mean study time is the only reason for higher exam scores. Other factors, such as prior knowledge or motivation, may also affect the results.</p>

<p>In short:</p>
<p><b>Correlation</b> - Variables are related.</p>
<p><b>Causation</b> - One variable causes a change in another.</p>

````python
plt.figure(figsize = (8, 6))
sns.heatmap(df.corr(numeric_only = True))
plt.show()
````

````python
plt.figure(figsize = (8, 6))
sns.heatmap(df.corr(numeric_only = True), annot = True, cmap = 'coolwarm')
plt.show()
````

<p>Read the heatmap by looking at the <b>correlation values</b> inside the cells:</p>

<ul>
    <li><b>1</b> - Variable compared with itself.</li>
    <li><b>0.68</b> - <code>total_bill</code> and <code>tip</code> have a <b>strong positive correlation</b>.</li>
    <li><b>0.60</b> - <code>total_bill</code> and <code>size</code> have a <b>moderate positive correlation</b>.</li>
    <li><b>0.49</b> - <code>tip</code> and <code>size</code> have a <b>moderate positive correlation</b>.</li>
</ul>

<p>In short: All three variables have a <b>positive relationship</b>, but none has a perfect relationship with another variable.</p>
