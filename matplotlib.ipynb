<h5>Subplot</h5>

<p><b>plt.subplot()</b> is used to display multiple plots in a single figure by dividing the figure into rows and columns.</p>

```bash
plt.subplot(rows, columns, index)
```

<p>Think of subplot like dividing one paper into multiple boxes. We decide how many rows and columns the paper has, then choose which box you want to draw in.</p>

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title("Plot 1")

plt.subplot(2, 2, 2)
plt.plot(x, y)
plt.title("Plot 2")

plt.subplot(2, 2, 3)
plt.scatter(x, y)
plt.title("Plot 3")

plt.subplot(2, 2, 4)
plt.bar(x, y)
plt.title("Plot 4")

plt.tight_layout()

plt.show()
````

<p>Here, <b>plt.tight_layout()</b> automatically adjusts the spacing between subplots, but it does not guarantee perfect padding in every situation.</p>

<p>What does it do?</p>
<p>It automatically adjusts the spacing between plots and labels so that everything fits properly.</p>
<p>Benefits</p>
<ul>
    <li>Titles don't overlap with other plots.</li>
    <li>X-axis labels don't overlap.</li>
    <li>Y-axis labels don't overlap.</li>
    <li>Subplots have better spacing.</li>
</ul>

<p>Increase the padding - </p>

```bash
plt.tight_layout(pad=2)
```

<ul>
    <li><strong>pad</strong> - Controls the padding around the entire figure.</li>
</ul>
<p>Control horizontal and vertical spacing - </p>

```bash
plt.tight_layout(h_pad=2, w_pad=2)
```

<ul>
    <li><strong>h_pad</strong> - Vertical padding between subplots.</li>
    <li><strong>w_pad</strong> - Horizontal padding between subplots.</li>
</ul>

<h5>Saving a Figure</h5>

<p><b>plt.savefig()</b> is used to save a Matplotlib figure as an image file.</p>

```bash
plt.savefig("filename.png")
```
<p>Can also in .jpg, .svg formats</p>

```bash
plt.savefig("my_plot.png", dpi=300, bbox_inches="tight")
```
<p>Here, </p>
dpi=300 - Saves the figure with higher resolution<br>
bbox_inches="tight" - Removes unnecessary extra whitespace around the figure

<p style="color:green;">Note: Use plt.savefig() before plt.show() to ensure the figure is saved correctly.</p>

````python
x = np.arange(11)
y = x**2

plt.subplot(2, 3, 1)
plt.plot(x, y, 'r-*')

plt.subplot(2, 3, 2)
plt.plot(y, x, 'g-*')

plt.subplot(1, 3, 3)
plt.plot(x, y, 'b--')

plt.tight_layout()
plt.savefig("my_plot.png", dpi = 300, bbox_inches = 'tight')
plt.show()
````

<p><b>plt.subplots()</b> is used to create a figure and one or more subplots at the same time.</p>
<p>It returns two objects:<br>
- <b>fig</b> - The entire figure<br>
- <b>ax</b> - The Axes (plotting area)<br>

Syntax - 

```bash
fig, ax = plt.subplots()
```

<b>plt.subplot()</b> - Creates or selects one subplot position at a time. To create multiple plots in a single figure, use <b>plt.subplot()</b> repeatedly.

<table>
  <tr>
    <th>plt.subplot()</th>
    <th>plt.subplots()</th>
  </tr>
  <tr>
    <td>Pyplot / Procedural style</td>
    <td>Object-Oriented (OO) style</td>
  </tr>
  <tr>
    <td>Creates or selects one subplot at a time</td>
    <td>Creates a Figure and one or more Axes together</td>
  </tr>
  <tr>
    <td>Uses <b>plt.plot()</b></td>
    <td>Uses <b>ax.plot()</b></td>
  </tr>
  <tr>
    <td>Uses <b>plt.title()</b></td>
    <td>Uses <b>ax.set_title()</b></td>
  </tr>
  <tr>
    <td>Uses <b>plt.xlabel()</b></td>
    <td>Uses <b>ax.set_xlabel()</b></td>
  </tr>
  <tr>
    <td>For multiple plots, <b>plt.subplot()</b> must be called repeatedly for each subplot</td>
    <td>All subplot Axes are created together using <b>plt.subplots()</b></td>
  </tr>
  <tr>
    <td>Simpler for basic use</td>
    <td>Better for managing multiple plots</td>
  </tr>
</table>

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

fig, ax = plt.subplots(2, 2)  # 2 rows, 2 cols
 
ax[0, 0].plot(x, y)  # First row, first column
ax[0, 1].scatter(x, y) # First row, second column
ax[1, 0].bar(x, y) # Second row, first column
ax[1, 1].plot(x, y)  # Second row, second column

plt.tight_layout()
plt.show()
````

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

fig, ax = plt.subplots(1, 2)

ax[0].plot(x, y)
ax[1].scatter(x, y)

plt.show()
````

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

fig, ax = plt.subplots(figsize = (10,8))   # figsize = (width, height) in inches
ax.plot(x,y, color = "red", linewidth = 0.5, linestyle = "-", marker = "o", markersize = 8, markerfacecolor = "blue", markeredgecolor = "green", markeredgewidth = 3)
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_title("Object Oriented Style Figure")
plt.show()
````

<p>Here, <b>figsize=(10, 8)</b> sets the size of the entire Matplotlib figure.</p>

````python
fig, ax = plt.subplots(2, 2, figsize=(10, 8))

ax[0, 0].plot(x, y)
ax[0, 1].scatter(x, y)
ax[1, 0].bar(x, y)
ax[1, 1].plot(x, y)

plt.tight_layout()
plt.show()
````

<h5>Legend in Matplotlib</h5>
<p>A legend is a box displayed on a graph that helps us identify what each line, marker, or plot represents.<p>
<p>For example, if a graph contains three lines:<br>
- Temperature<br>
- Ice Cream Sales<br>
- Power Cuts<br>

The legend tells us which line belongs to which data.</p>


````python
x = [1, 2, 3, 4, 5]

temperature = [25, 27, 28, 30, 32]
sales = [100, 120, 150, 180, 220]
power_cuts = [5, 4, 6, 3, 2]

plt.plot(x, temperature, label="Temperature")
plt.plot(x, sales, label="Ice Cream Sales")
plt.plot(x, power_cuts, label="Power Cuts")

plt.legend()
plt.show()
````

<p>Here, <b>label</b> defines the legend name and <b>plt.legend()</b> shows the legend. The <b>plt.legend()</b> function displays a legend on the plot. A legend explains what each color, marker, or line represents.</p>
<p>Syntax - </p>

```bash
plt.legend(loc='position')
```
<p>The <b>loc</b> parameter specifies where the legend will be placed.</p>

<p><b>bbox_to_anchor</b></p>

<p>The <b>bbox_to_anchor</b> parameter let us place the legend at a custom position. It is commonly used to move the legend <b>outside the plot</b>.</p>

```bash
plt.legend(loc='upper left', bbox_to_anchor=(1.02, 1))
```

<p>How It Works?</p>
<ul>
<li><b>loc</b> chooses which corner of the legend is used as the reference point.</li>
<li><b>bbox_to_anchor=(x, y)</b> specifies where that reference point should be placed.</li>
</ul>

<p style = "color: red;">Note : Choose the bbox_to_anchor values as needed....</p>

````python
x = [1, 2, 3, 4, 5]

temperature = [25, 27, 28, 30, 32]
sales = [100, 120, 150, 180, 220]
power_cuts = [5, 4, 6, 3, 2]

plt.plot(x, temperature, label="Temperature")
plt.plot(x, sales, label="Ice Cream Sales")
plt.plot(x, power_cuts, label="Power Cuts")

plt.legend(loc = 'upper center' , bbox_to_anchor = (0.5, -0.15))
plt.show()
````
