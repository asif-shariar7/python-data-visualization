<h5>Time Series Analysis</h5>

<p>
A time series is data collected or recorded over time. For example:
</p>

<ul>
<li>Daily temperature</li>
<li>Monthly sales</li>
<li>Yearly population</li>
<li>Hourly stock prices</li>
<li>Daily website visitors</li>
</ul>

<p>
The main goal is to understand how the data changes over time and sometimes use past data to predict the future.
</p>


<h5>Time Series Data Visualization</h5>

<p>
Time Series Data Visualization means using graphs and charts to visualize data that changes over time. 
The main purpose is to understand trends, patterns, seasonality, and changes over time.
</p>

<p><b>Common Visualizations:</b></p>

<ol>

<li>
<b>Line Chart</b>
<p>
It connects data points in time order and helps to see how values change over time.
Example: Monthly sales from January to December.
<br>
Best for: Trends and changes over time
</p>
</li>

<li>
<b>Bar Chart</b>
<p>
It can be used when we want to compare values across different time periods.
Example: Comparing sales for each month.
<br>
Best for: Comparing values across time periods
</p>
</li>

<li>
<b>Moving Average Plot</b>
<p>
A moving average is often added to a time series plot to make the data smoother.
</p>

<p>
For example:
</p>

<ol>
<li><b>Actual Sales:</b> The real sales numbers. They can go up and down every month.</li>
<li><b>Moving Average:</b> Takes the average of nearby sales values, so the line becomes smoother and makes the general direction of sales easier to see.</li>
</ol>

<p>
Best for: Reducing noise and identifying trends.
</p>
</li>

<li>
<b>Seasonal Decomposition Plot</b>
<p>
This visualization breaks a time series into separate components:
</p>
<b>Observed</b> - Original time series data, which is decomposed into 3 components:
<ul>
<li><b>Trend</b> - Long-term movement</li>
<li><b>Seasonal</b> - Repeating pattern</li>
<li><b>Residual</b> - Remaining random variation</li>
</ul>

<p>
Best for: Understanding the different components of a time series.
</p>
</li>

<li>
<b>Area Chart</b>
</li>

</ol>


<p>
When analyzing a time series graph, we usually look for :
</p>

<ol>
<li><b>Trend</b> - Is the data increasing or decreasing?</li>
<li><b>Seasonality</b> - Is there a repeating pattern?</li>
<li><b>Cyclic behavior</b> - Are there long-term fluctuations?</li>
<li><b>Outliers</b> - Are there unusual values?</li>
<li><b>Volatility</b> - How much does the data fluctuate?</li>
<li><b>Change points</b> - Did the behavior suddenly change?</li>
</ol>


<p>
<b>Example - </b>
</p>

<p>
Imagine a graph of a shop's monthly sales:
</p>

<ul>
<li><b>Raw Time Series</b> - Shows actual sales over time</li>
<li><b>Moving Average</b> - Makes the overall trend easier to see</li>
<li><b>Seasonal Decomposition</b> - Separates the data into Trend + Seasonal + Residual</li>
</ul>

<h5>Noise in Time Series Data</h5>
<p>
Noise means random, unwanted, or unexpected changes in the data that do not represent a meaningful pattern.
</p>
<p><b>Example:</b></p>
<p>
Suppose a shop's sales are generally increasing:
</p>
<p>
100 -> 110 -> 120 -> 130 -> 140
</p>
<p>
But the actual sales are:
</p>
<p>
100 -> 108 -> 125 -> 127 -> 145
</p>
<p>
The small unexpected ups and downs are noise.
</p>

<p>
The random fluctuations around the general trend are called noise.
</p>
<p>Why is noise important?</p>
<p>
Noise can make it difficult to see the real trend or pattern in data.
</p>
<p>
That's why we often use a Moving Average to smooth the data:
</p>
<ul>
<li><b>Actual Data</b> -> Contains noise and fluctuations</li>
<li><b>Moving Average</b> -> Reduces some noise and makes the trend easier to see</li>
</ul>

<hr>

<h5>Moving Average</h5>
<p>
A Moving Average calculates the average of a fixed number of recent data points and moves this calculation forward over time. It is mainly used to smooth noisy data and see the overall trend more clearly.
</p>

<p><b>Example:</b></p>
<p>
Suppose daily sales are:
</p>
<p>
10, 20, 30, 40, 50
</p>
<p>
Using a 3-day moving average:
</p>
<ul>
<li>First average = (10 + 20 + 30) / 3 = 20</li>
<li>Next average = (20 + 30 + 40) / 3 = 30</li>
<li>Next average = (30 + 40 + 50) / 3 = 40</li>
</ul>
<p>
So the moving average becomes:
</p>
<p>
20, 30, 40
</p>
<p><b>Why use it?</b></p>
<p>
It removes some short-term fluctuations and makes the underlying trend easier to see.
</p>

<p>Why Use Moving Averages?</p>
<table>
    <tr>
        <th>Purpose</th>
        <th>Benefit</th>
    </tr>
    <tr>
        <td>Reduce noise in data</td>
        <td>Makes trends easier to see</td>
    </tr>
    <tr>
        <td>Reveal seasonal patterns</td>
        <td>Useful in time series (like weather)</td>
    </tr>
    <tr>
        <td>Improve readability</td>
        <td>Helps non-technical users spot trends</td>
    </tr>
</table>

<h5>Seasonal Decomposition</h5>
<p>
Seasonal Decomposition is a method that separates a time series into:
</p>
<p>
Observed Data -> Trend + Seasonality + Residual
</p>
<p>
For example, imagine monthly sales data.
</p>
<p>
The decomposition can show:
</p>
<ul>
<li><b>Observed:</b> Actual sales</li>
<li><b>Trend:</b> Sales are generally increasing</li>
<li><b>Seasonal:</b> Sales increase every December</li>
<li><b>Residual:</b> Unexpected changes</li>
</ul>
<p>Why It's Useful?</p>
<table>
    <tr>
        <th>Purpose</th>
        <th>Insight Example</th>
    </tr>
    <tr>
        <td>Detect seasonality</td>
        <td>"Temps rise every June – it's predictable!"</td>
    </tr>
    <tr>
        <td>Isolate long-term trend</td>
        <td>"Temps are rising year over year – climate change?"</td>
    </tr>
    <tr>
        <td>Separate noise from signal</td>
        <td>"Random heatwave isn't a trend"</td>
    </tr>
</table>

<hr>

<h5>Trend</h5>
<p>
The Trend shows the long-term direction of the data.
</p>
<p>
The data may be:
</p>
<ul>
<li>Increasing</li>
<li>Decreasing</li>
<li>Relatively stable</li>
</ul>
<p><b>Example:</b></p>
<p>
A company's sales:
</p>
<p>
100 -> 120 -> 140 -> 160 -> 180
</p>
<p>
The overall trend is increasing.
</p>
<p>
The individual values may go up and down, but the trend shows the general long-term direction.
</p>

<h5>Seasonality</h5>
<p>
Seasonality is a pattern that repeats at a fixed and predictable interval.
</p>
<p><b>Examples:</b></p>
<ul>
<li>Ice cream sales increase every summer.</li>
<li>Umbrella sales increase during the rainy season.</li>
<li>Online shopping increases every December.</li>
<li>Electricity usage may follow a daily pattern.</li>
</ul>
<p>
For monthly data, yearly seasonality often has a 12-month period.
</p>
<p>
For daily data, weekly seasonality may have a 7-day period.
</p>

<h5>Residual</h5>
<p>
After removing the trend and seasonality, some variation is still left. This remaining part is called the Residual or Error. It represents things that the model could not explain, including random fluctuations and unexpected events.
</p>

<p><b>Example:</b></p>
<p>
A shop normally sells 100 products, but suddenly sells only 50 because of a power outage.
</p>
<p>
This unexpected change may appear in the residual component.
</p>

<table>
    <tr>
        <th>Component</th>
        <th>Meaning</th>
    </tr>
    <tr>
        <td>Trend</td>
        <td>Long-term direction (e.g., rising global temperatures)</td>
    </tr>
    <tr>
        <td>Seasonal</td>
        <td>Repeating pattern (e.g., hot summers, cool winters every year)</td>
    </tr>
    <tr>
        <td>Residual</td>
        <td>What's left over (random noise or unexplained variation)</td>
    </tr>
</table>

<p>There are two models:</p>

<ul>
    <li><b>additive</b> — if data = trend + seasonality + noise</li>
    <li><b>multiplicative</b> — if data = trend × seasonality × noise</li>
</ul>

<h5>Area Chart - </h5>

```bash 
plt.fill_between(x, y, color='color', alpha=value, label='label')
```
<p>Here - </p>

```bash
plt.fill_between(x, y)  #Fills the area between the x-axis and the y-values.
```

<h5>Seasonal decomposition</h5>
<p>Import - </p>

```bash
from statsmodels.tsa.seasonal import seasonal_decompose
```
<p>Here, </p>
<ul>
<li><b>statsmodels</b> - A Python library for statistical analysis and time series modeling.</li>
<li><b>tsa</b> - Stands for <b>Time Series Analysis</b>.</li>
<li><b>seasonal</b> - Module containing seasonal analysis functions.</li>
<li><b>seasonal_decompose</b> - Function used to decompose a time series into <b>Trend</b>, <b>Seasonal</b>, and <b>Residual</b> components.</li>
</ul>

<p>Installation - </p>

```bash
pip install statsmodels
```
<p>Syntax - </p>

```bash
result = seasonal_decompose(time_series, model='additive', period=n)
```

<p><b>seasonal_decompose()</b> - The function used to decompose a time series.</p>

<p><b>time_series</b> - The column containing the time series data.</p>
<p>Example: df['Avg_Temperature']</p>

<p><b>model</b> - Specifies the decomposition model (additive or multiplicative).</p>
<p>Additive Model - Used when the seasonal variation remains approximately constant over time.<br>
Formula: Observed = Trend + Seasonal + Residual</p>

<p>Multiplicative Model - Used when seasonal variation changes proportionally with the trend.<br>
Formula: Observed = Trend × Seasonal × Residual</p>

<p><b>period</b> - Specifies the length of one complete seasonal cycle.</p>

<tr>
<th>Data Frequency</th>
<th>Period</th>

<td>7</td>

<td>12</td>

<p>Accessing the Components - <br>
trend = result.trend<br>
seasonal = result.seasonal<br>
residual = result.resid</p>

<p>Plotting the Components - <b>result.plot()</b> which displays four plots:
<ol>
<li><b>Observed (Original Data)</b></li>
<li><b>Trend</b></li>
<li><b>Seasonal</b></li>
<li><b>Residual</b></li>
</ol></p>

````python
result = seasonal_decompose(df['Avg_Temperature'], model = 'additive', period = 12)
result.plot()
plt.suptitle("Seasonal Decomposition of temperature", fontsize=16)
plt.tight_layout()
plt.show()
````
