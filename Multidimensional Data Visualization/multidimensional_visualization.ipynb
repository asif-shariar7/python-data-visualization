<h5>What is Multidimensional Data?</h5>

<p>
Multidimensional data means data that has many <b>features/columns</b>.
</p>

<p>
For example, the <b>Wine dataset</b> contains information about wine using
<b>13 features</b>, such as:
</p>

<ul>
    <li>Alcohol</li>
    <li>Malic acid</li>
    <li>Magnesium</li>
    <li>Color intensity</li>
    <li>Flavanoids</li>
    <li>Proline</li>
    <li>etc.</li>
</ul>

<p>
So, one wine sample can look like:
</p>

```bash
[13.0, 2.0, 2.3, 15.6, ..., 1.2]
```

<p>
This means one sample has many values/features.
</p>

<p>The Problem : </p>

<p>We can easily visualize : </p>

<ol>
    <li><b>1 feature - 1D</b></li>
    <li><b>2 features - 2D</b></li>
    <li><b>3 features - 3D</b></li>
</ol>

<p>
But it is very difficult to visualize <b>13 features</b> at the same time.
</p>

<p>
Therefore, we need a technique to <b>reduce the dimensions</b> for visualization.
</p>

<p>
That is where <b>t-SNE</b> is useful.
</p>

<h5>What is t-SNE?</h5>

<p>
<b>t-SNE</b> stands for :
</p>

<blockquote>
    <b>t-distributed Stochastic Neighbor Embedding</b>
</blockquote>

<p>
It is a technique used to <b>reduce high-dimensional data into 2D or 3D</b>
so that we can visualize it.
</p>

<p>The important idea is : </p>

<blockquote>
    <b>t-SNE tries to keep similar data points close together.</b>
</blockquote>

<p>
t-SNE can reduce high-dimensional data, such as the <b>13 features</b> of the
Wine dataset, into <b>2 or 3 dimensions</b> while trying to preserve
similarities and clusters of the original data.
</p>


<p>Imagine we have <b>100 students</b>.</p>

<p>Each student has:</p>

<ul>
    <li>Math marks</li>
    <li>English marks</li>
    <li>Physics marks</li>
    <li>Chemistry marks</li>
    <li>Biology marks</li>
    <li>Attendance</li>
    <li>Assignment marks</li>
    <li>etc.</li>
</ul>

<p>
Maybe there are <b>10 features</b> for each student.
</p>

<p>
It is difficult to make a graph using all 10 features.
</p>

<p>
t-SNE can convert : 10 features -> t-SNE -> 2 features
</p>


<p>
Then we can make a simple <b>2D scatter plot</b>.
</p>

<p>Main Goal of t-SNE : </p>

<p>
The main goal of t-SNE is <b>visualization</b>.
</p>

<p>It helps us find:</p>

<ul>
    <li>Groups</li>
    <li>Clusters</li>
    <li>Similar data points</li>
    <li>Different data points</li>
    <li>Hidden patterns</li>
</ul>

````python
from sklearn.datasets import load_wine
from sklearn.manifold import TSNE
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
data = load_wine()
data
````

````python
print(data.keys())
````

````python
data.feature_names
````

````python
data.data
````

<p>The input is : <b>data.data</b></p>
<p>It contains : 178 samples x 13 features</p>
<p>After t-SNE, we will get : 178 samples × 2 dimensions(if n_components = 2)</p>

````python
df = pd.DataFrame(data.data, columns = data.feature_names)
df.head()
````

````python
df.shape
````

````python
df = pd.DataFrame(data.data, columns = data.feature_names)
df['target'] = data.target
df.head()
````

````python
df.shape
````

````python
tsne = TSNE(n_components = 2, random_state = 42, perplexity = 30)
tsne_result = tsne.fit_transform(data.data)
````

<p>Here, </p>
<p>
<b>n_components = 2</b> means : We want the final result to have <b>2 dimensions</b>.</p>

<p>Originally, the Wine dataset has : 13 features</p>

<p>After applying t-SNE:</p>

<p>13 features -> t-SNE -> 2 dimensions</p>

<p>
Now we can plot the result using an <b>X-Y graph</b>.</p>


<p><b>random_state = 42</b></p>

<p>
t-SNE involves some <b>randomness</b>. Therefore, running t-SNE multiple
times can give slightly different results.
</p>

<p>
We use : random_state=42. to make the result more <b>reproducible</b>.</p>


<p>
Simple meaning : <b>random_state = make the random result repeatable</b>
</p>

<p>
The number <b>42</b> is not special. We can use another number, such as : random_state = 10
</p>

<p>
The important thing is to use the <b>same fixed number</b> when we want
repeatable results.
</p>

<p><b>perplexity = 30</b></p>

<p>
Perplexity controls the neighborhood size. In simple words, it controls how many nearby data points t-SNE pays
attention to when finding similarities. A common range is approximately <b>5–50</b>. For basic visualization, we do not need to understand the mathematical
formula behind perplexity.
</p>


<p><b>More about Perplexity and Neighborhood</b></p>

<p><b>Neighborhood : </b> It means the nearby or similar data points around one particular data point</p>

<p>
<b>Perplexity</b> controls the <b>approximate size of the neighborhood</b>
that t-SNE considers when finding similarities.
</p>

<p>For example : TSNE(perplexity = 30)</p>


<p>
This means t-SNE considers a neighborhood of <b>roughly 30 nearby points</b>
when understanding the similarity of each point. It does not mean exactly 30 nearest points. It is an
approximate way to think about the neighborhood size
</p>

<p>So, 
Neighborhood = nearby points. Perplexity = controls the size of that neighborhood
</p>

<ul>
    <li><b>perplexity = 10</b> - smaller neighborhood</li>
    <li><b>perplexity = 30</b> - medium neighborhood</li>
    <li><b>perplexity = 50</b> - larger neighborhood</li>
</ul>

<p><b>Example :</b> Imagine you are looking at one student. There are <b>100 students</b> around you.
</p>

<p>t-SNE needs to decide : <b>"Which students are similar to this student?"</b></p>

<p>
It does not need to pay equal attention to all 100 students.
</p>

<p>With : perplexity = 30. we can <b>roughly think of it</b> as t-SNE paying attention to about
<b>30 nearby/neighborhood points</b> when deciding which points are similar.</p>

<p>What does <b>fit_transform()</b> do?</p>

<blockquote>
    It learns the relationships between the data points and transforms them
    into a new 2D space.
</blockquote>


````python
tsne_result
````

<p>What Does <b>tsne_result</b> Contain?</p>

<p>
<b>tsne_result</b> contains <b>two new values for every wine sample</b>.
</p>

<p>For example:</p>

```bash
[24.5, -3.2]
[21.1, -4.7]
[-15.2, 8.4]
...
```

<p>
These two values can be used as :
</p>

<ul>
    <li><b>X coordinate</b></li>
    <li><b>Y coordinate</b></li>
</ul>

<p>
For example, originally :
</p>

<p>
Wine 1 -> 13 feature values. After t-SNE -> Wine 1 : [24.5, -3.2]
</p>


<p>
So t-SNE changes the 13-dimensional data into a 2-dimensional coordinate.
</p>



<p>What Do These Numbers Mean?</p>

<p>
Suppose you see : X = 24.5 , Y = -3.2
</p>


<p>
<b>Do not think : X = Alcohol and Y = Magnesium</b>
</p>

<p>
That is <b>wrong</b>.
</p>

<p>
The t-SNE X and Y axes do <b>not</b> represent the original features.
They are newly created coordinates.
</p>

<p>
The most important thing to look at is :
</p>

<blockquote>
    <b>Which data points are close to each other?</b>
</blockquote>

<p>
If points are close together, they are considered <b>similar</b> in the
high-dimensional data.
</p>

````python
df['tsne-2d-one'] = tsne_result[:, 0]   # [:, 0] means take the first value from every row.
df['tsne-2d-two'] = tsne_result[:, 1]   # [:, 1] means take the second value from every row.
df.head()
````

<p>Here, The : (colon) means : Take all rows.</p>

````python
plt.figure(figsize=(10,7))
sns.scatterplot(x='tsne-2d-one', y='tsne-2d-two', hue='target', data=df, palette='deep')
plt.title('t-SNE on Wine dataset')
plt.show()
````

<p>So, We have 13 features, which are difficult to visualize together. t-SNE uses these 13 features and converts each data point into 2 new coordinates. We then use these coordinates to visualize the 3 target classes and see their similarities, differences, and clusters.

<p>Note : What is a Cluster?</p>
<p>
A <b>cluster</b> is a group of data points that are
<b>close or similar to each other</b>.
</p>

<h5>Observation from the t-SNE Plot</h5>

<ul>
<li>The Wine dataset has <b>13 features</b>, which t-SNE reduces to <b>2D</b>.</li>
<li>The three wine classes form <b>different groups</b>.</li>
<li><b>Class 0</b> (blue) forms a tight and well-separated group.</li>
<li><b>Class 1</b> (orange) and <b>Class 2</b> (green) are mostly separate, but they overlap a little.</li>
<li>This means the original <b>13 features</b> contain useful information for distinguishing the wine classes.</li>
<li>The overlap between <b>Class 1</b> and <b>Class 2</b> suggests that they are more similar to each other than to <b>Class 0</b>.</li>
</ul>
